```python
import pandas as pd
```

```python
df = pd.read_excel(
    r"C:\Users\Bala murukan\Downloads\New folder\EV_Intelligencee_platform\01_Raw_Data\CMO-Historical-Data-Monthly.xlsx",
    sheet_name="Monthly Prices",
    header=None,
    engine="openpyxl"
)
```

```python
df.head(10)
```

```python
print(df.iloc[4:10, 0:3])
```

```python
# Skip the header rows (usually first 6 rows)
df = df.iloc[6:].copy()

# Keep only first two columns (Date and Crude oil average)
df = df.iloc[:, [0, 1]]

df.columns = ["Date", "Crude_Oil_Price"]

df.head(10)
```

```python
df["Date"] = df["Date"].astype(str)
df = df[df["Date"].str.contains("M", na=False)]   # Keep only monthly rows

df["Year"] = df["Date"].str[:4].astype(int)
df["Crude_Oil_Price"] = pd.to_numeric(df["Crude_Oil_Price"], errors="coerce")

df = df.dropna(subset=["Crude_Oil_Price"])
df.head()
```

```python
annual = df.groupby("Year")["Crude_Oil_Price"].mean().reset_index()
annual = annual[(annual["Year"] >= 2015) & (annual["Year"] <= 2024)]

print(annual)
print("Final rows:", len(annual))
```

```python
annual.to_csv(
    r"C:\Users\Bala murukan\Downloads\New folder\EV_Intelligencee_platform\02_Cleaned_Data\cleaned_crude_oil.csv",
    index=False
)

print("Crude Oil dataset cleaned successfully!")
print("Final rows:", len(annual))
```
