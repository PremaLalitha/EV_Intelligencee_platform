```python
df = df[["Country Name", "Country Code", "Indicator Name", "Indicator Code"] + [str(y) for y in range(2015, 2025)]]
df.head()
```

```python
pop_long = df.melt(
    id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
    var_name="Year",
    value_name="Population"
)

pop_long.head()
```

```python
pop_long.shape
```

```python
print(pop_long.isnull().sum())
```

```python
pop_long = pop_long.dropna(subset=["Population"])
print("After removing missing values:", pop_long.shape)
```

```python
pop_long = pop_long.rename(columns={
    "Country Name": "Country",
    "Country Code": "Country_Code",
    "Indicator Name": "Indicator_Name",
    "Indicator Code": "Indicator_Code"
})
```

```python
regions_to_remove = [
    'Africa Eastern and Southern', 'Africa Western and Central', 'Arab World',
    'Caribbean small states', 'Central Europe and the Baltics', 'Channel Islands',
    'East Asia & Pacific (excluding high income)', 'Early-demographic dividend',
    'East Asia & Pacific', 'Europe & Central Asia (excluding high income)',
    'Europe & Central Asia', 'Euro area', 'European Union',
    'Heavily indebted poor countries (HIPC)', 'High income',
    'IBRD only', 'IDA & IBRD total', 'IDA total', 'IDA blend', 'IDA only',
    'Latin America & Caribbean (excluding high income)', 'Latin America & Caribbean',
    'Least developed countries: UN classification', 'Low income',
    'Lower middle income', 'Low & middle income', 'Late-demographic dividend',
    'Middle East, North Africa, Afghanistan & Pakistan',
    'Middle East, North Africa, Afghanistan & Pakistan (excluding high income)',
    'Middle income', 'North America', 'OECD members', 'Other small states',
    'Pre-demographic dividend', 'Pacific island small states',
    'Post-demographic dividend', 'South Asia',
    'Sub-Saharan Africa (excluding high income)', 'Sub-Saharan Africa',
    'Small states', 'East Asia & Pacific (IDA & IBRD countries)',
    'Europe & Central Asia (IDA & IBRD countries)',
    'Latin America & the Caribbean (IDA & IBRD countries)',
    'Middle East, North Africa, Afghanistan & Pakistan (IDA & IBRD)',
    'South Asia (IDA & IBRD)', 'Sub-Saharan Africa (IDA & IBRD countries)',
    'Upper middle income', 'World'
]

pop_long = pop_long[~pop_long["Country"].isin(regions_to_remove)]

print("After removing regions:", pop_long.shape)
```

```python
pop_long.to_csv(
    r"C:\Users\Bala murukan\Downloads\New folder\EV_Intelligencee_platform\02_Cleaned_Data\cleaned_population.csv",
    index=False
)

print("Population dataset cleaned successfully!")
print("Final rows:", len(pop_long))
```
