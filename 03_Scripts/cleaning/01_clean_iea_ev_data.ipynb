```python
df["region_country"]=df["region_country"].str.strip()
```

```python
df["category"]=df["category"].str.strip()
```

```python
df["region_country"] = df["region_country"].replace({
    "Viet Nam": "Vietnam",
    "Korea": "South Korea",
    "Turkiye": "Turkey",
    "USA": "United States",
    "Czech Republic": "Czechia"
})

print(df["region_country"].unique())
```

```python
df["powertrain"].unique()
```

```python
df["mode"].unique()
```

```python
# Keep only EV sales and EV stock
df = df[df["parameter"].isin(["EV sales", "EV stock"])]
```

```python
# Keep only Historical data
df = df[df["category"] == "Historical"]
```

```python
# Keep years 2015 to 2024
df = df[df["year"].between(2015, 2024)]
```

```python
df.shape
```

```python
df["year"] = df["year"].astype(int)
df["value"] = pd.to_numeric(df["value"])
```

```python
df.rename(columns={
    "region_country": "Country",
    "category": "Category",
    "parameter": "Parameter",
    "mode": "Mode",
    "powertrain": "Powertrain",
    "year": "Year",
    "unit": "Unit",
    "value": "Value",
    "Aggregate group": "Aggregate_Group"
}, inplace=True)
```

```python
# List of regions to remove
regions_to_remove = [
    "Asia Pacific",
    "Europe",
    "EU27",
    "North America",
    "Central and South America",
    "Rest of the world",
    "Middle East and Caspian",
    "Africa"
]

# Keep only real countries + World
df = df[~df["Country"].isin(regions_to_remove)]

print("After removing regions:")
print("Total rows:", len(df))
print(df["Country"].unique())
```
