```python
df["Date"] = pd.to_datetime(df["Date"])

# Keep data from 2015 to 2024
df = df[(df["Date"].dt.year >= 2015) & (df["Date"].dt.year <= 2024)]

print("Shape after filtering years:", df.shape)
```
